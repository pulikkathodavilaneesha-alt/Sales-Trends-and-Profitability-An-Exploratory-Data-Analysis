```python
df[['Sales','Quantity','Profit']].boxplot()
```

```python
# Bivariate Analysis
df.plot.scatter(x='Sales',y='Profit',title = 'Sales vs Profit')
```

```python
df["Sales"].corr(df["Profit"])
```

```python
df["Month"] = pd.to_datetime(df["Order Date"]).dt.to_period('M')
monthly_sales = df.groupby("Month")["Sales"].sum()
monthly_sales.plot(kind = 'line', x = 'Month', y = 'Sales', title='Sales trend over time')
```

```python
#Multivariate Analysis
sns.heatmap(df[['Sales','Profit','Quantity','Discount']].corr(numeric_only = True),annot = True)
plt.rcParams['figure.figsize']=(10,15)
plt.show()
```

```python
sns.pairplot(df[['Quantity','Sales','Profit']])
```

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
category_sales.plot(kind='barh',title = 'Categorial Sales',figsize =(12,10)
print('Sales:', category_sales)
```

```python
profit_performance = df.groupby('Region')['Profit'].mean()
profit_performance.plot(kind='bar',title='Profit performance across regions')
print(profit_performance)
```

```python
segment_sales = df.groupby('Segment')['Sales'].sum()
highest_segment = segment_sales.idxmax()
highest_segment_sales = segment_sales.max()
print("Highest Sales segment:", highest_segment)
print("Sales:", highest_segment_sales)
```
